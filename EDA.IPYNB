# Breast Cancer EDA and Baseline Modeling

Quick checks on the Wisconsin dataset before training: feature ranges and class balance.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
dataset = load_breast_cancer(as_frame=True)
df = dataset.data.copy()
target = dataset.target.copy()
df['target'] = target
df.head()
```

```python
df.describe().T[['mean', 'std', 'min', 'max']].head(10)
```

```python
target_counts = df['target'].value_counts().rename(index={0: 'malignant', 1: 'benign'})
target_counts
```
